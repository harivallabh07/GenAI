```python
from tensorflow.keras.models import load_model
import pickle
import numpy as np
import pandas as pd
```

```python
## Loading the model, Ohe_encoder, label_gender and Scalar Pickle files
model=load_model('model_1.h5')

with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender=pickle.load(file)
with open('ohe_geo_encoder.pkl','rb') as file:
    ohe_geo_encoder=pickle.load(file)
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
#Example data
input_data={
    'CreditScore':650,
    'Geography':'France',
    'Gender':'Female',
    'Age':45,
    'Tenure':1,
    'Balance':70000,
    'NumOfProducts':2,
    'HasCrCard':5,
    'IsActiveMember':1,
    'EstimatedSalary':30000
}
```

```python
geo_encoded=ohe_geo_encoder.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded,columns=ohe_geo_encoder.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_data=pd.DataFrame([input_data])
input_data=pd.concat([input_data.drop('Geography',axis=1),geo_encoded_df],axis=1)
input_data['Gender']=label_encoder_gender.transform(input_data['Gender'])
input_data
```

```python
#scaling
input_scaled=scaler.transform(input_data)
input_scaled
```

```python
predicted=model.predict(input_scaled)
predicted
```

```python
if predicted>0.5:
    print(f"They'll be likely to leave the job since the probability is {predicted*100}")
else:
    print(f'No, they will continue working ')
```
