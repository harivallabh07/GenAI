# Data Preprocessing and Cleaning

```python
import pandas as pd
data=pd.read_csv('Churn_Modelling.csv')
data.head()
```

```python
data = data.drop(
    columns=['RowNumber', 'CustomerId', 'Surname'],
    errors='ignore'
)

data.head()
```

```python
#Encode all the Labeled data to categorical data like gender-> use label encoder
from sklearn.preprocessing import LabelEncoder
label_encoder_gender=LabelEncoder()
```

```python
data['Gender']=label_encoder_gender.fit_transform(data['Gender'])
data
```

```python
#since there are different geographical names -> better to use OHE
from sklearn.preprocessing import OneHotEncoder
ohe_geo_encoder=OneHotEncoder()
geo=ohe_geo_encoder.fit_transform(data[['Geography']])
geo
```

```python
ohe_geo_encoder.get_feature_names_out(['Geography'])
```

```python
geo_encoded_df=pd.DataFrame(geo.toarray(),columns=ohe_geo_encoder.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
data=pd.concat([data.drop('Geography', axis=1),geo_encoded_df],axis=1)
```

```python
data.head()
```

```python
#Saving the encoded valuses in the pickle files, so that they are saved and we can work on other part of the memory and dataset.
# Gender -> Label Encoder
#  Geography -> OHE
import pickle
with open('label_encoder_gender.pkl','wb') as file:
    pickle.dump(label_encoder_gender,file)
```

```python
with open('ohe_geo_encoder.pkl','wb') as file:
    pickle.dump(ohe_geo_encoder,file)
```

```python
#Divide teh dataset into Independent and Depedent Feature
data.head()
```

```python
#X-independent, y-depedent
X=data.drop(['Exited'],axis=1)
y=data['Exited']
```

```python
X.head(2)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=6)
```

```python
#scaling the data
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
#we'll save this Scalar data in the pickle file
with open('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

# ANN implementation

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

```python
model=Sequential([
    Dense(80,activation='relu',input_shape=(X_train.shape[1],)),#HL-1 - Connected to input layer
    Dense(40,activation='relu'),#HL-2
    Dense(1,activation='relu')#output layer
]
)
```

```python
model.summary()
```

```python
opt=tf.keras.optimizers.Adam(learning_rate=0.001)
loss=tf.keras.losses.BinaryCrossentropy()
loss
```

```python
#TO train the model we need to compile itfirst
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy']) 
```

```python
# set the TensorBoard 
log_dir='log/fit/'+datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorlfow_callbacks=TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
## setup the EarlyStopping
early_stopping_callback=EarlyStopping(monitor='val_loss',patience=20,restore_best_weights=True)
```

# Model Training

```python
history=model.fit(
    X_train,y_train,validation_data=(X_test,y_test),epochs=60,
    callbacks=[tensorlfow_callbacks,early_stopping_callback]
)
```

```python
model.save('model_1.h5')
```
